# Dataset Cleaning Pipeline

This notebook loads an Excel dataset, cleans it, and saves the result as a CSV file.

**Steps:**
1. Import libraries
2. Load dataset
3. Standardize column names
4. Inspect data
5. Validate required column
6. Clean data
7. Save output
8. Run the pipeline

## 1. Import Libraries

```python
import pandas as pd
from pathlib import Path
```

## 2. Load Dataset

Reads the raw dataset from an Excel file into a DataFrame.

```python
def load_dataset(file_path):
    file_path = Path(file_path)
    df = pd.read_excel(file_path, engine="openpyxl")
    return df
```

## 3. Standardize Column Names

Converts all column names to lowercase and strips whitespace.

```python
def standardize_columns(df):
    df.columns = df.columns.str.lower().str.strip()
    return df
```

## 4. Inspect Data

Quick look at the dataset: shape, first rows, structure, and missing values.

```python
def inspect_dataset(df):
    print("Shape:", df.shape)
    print(df.head())
    df.info()
    print(df.isnull().sum())
```

## 5. Validate Required Column

Ensures all required columns exist before proceeding.

```python
def validate_columns(df, required_cols):
    for col in required_cols:
        if col not in df.columns:
            raise ValueError(f"Missing required column: {col}")
```

## 6. Clean Data

Drops missing values in the `phrase` column, normalizes the text, and removes duplicate rows.

```python
def clean_data(df):
    df = df.dropna(subset=["phrase"])

    df["phrase"] = (
        df["phrase"]
        .astype(str)
        .str.lower()
        .str.strip()
    )

    df = df.drop_duplicates()
    return df
```

## 7. Save Output

Saves the cleaned DataFrame to a CSV file, creating parent folders if needed.

```python
def save_dataset(df, save_path):
    save_path = Path(save_path)
    save_path.parent.mkdir(parents=True, exist_ok=True)

    df.to_csv(save_path, index=False, encoding="utf-8-sig")

    print("Saved:", save_path)
```

## 8. Full Pipeline

Combines all steps into a single function.

```python
def clean_dataset(file_path, save_path):
    df = load_dataset(file_path)
    df = standardize_columns(df)

    inspect_dataset(df)

    validate_columns(df, ["phrase"])

    df = clean_data(df)

    save_dataset(df, save_path)

    return df
```
